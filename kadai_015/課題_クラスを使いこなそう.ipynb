```python
# Humanクラスを定義する
class Human:
    # 名前と年齢を初期設定する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を標準出力するメソッドを定義する
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)


# Humanクラスのインスタンスを変数に代入する
person = Human("侍太郎", 25)

# メソッドを呼び出して情報を表示する
person.printinfo()
```
