```python

# Humanクラスを定義する
class Human:
    # 名前と年齢を設定する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を表示する
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)

    # 成人かどうかを判定する（20歳以上を成人とする）
    def check_adult(self):
        if self.age >= 20:
            print(self.name, "さんは大人です。")
        else:
            print(self.name, "さんは大人ではありません。")


# Humanクラスのインスタンスを複数生成してリストに追加する
humans = [
    Human("太郎", 25),
    Human("花子", 18),
    Human("次郎", 20)
]

# リストの要素数分だけcheck_adultメソッドを呼び出す
for human in humans:
    human.check_adult()
```
