```python
# Humanクラスを定義する
class Human:
    # 名前と年齢を初期化する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 20歳以上かどうかを判定するメソッド
    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}さんは大人です。")
        else:
            print(f"{self.name}さんは大人ではありません。")


# 人物の情報をリストとディクショナリで管理する
people = [
    {"name": "侍太郎", "age": 25},
    {"name": "侍花子", "age": 18},
    {"name": "侍次郎", "age": 20},
    {"name": "侍三郎", "age": 17}
]

# 繰り返し処理で人物ごとに判定する
for person in people:
    # Humanクラスのインスタンスを作成する
    human = Human(person["name"], person["age"])

    # 大人かどうかを判定して出力する
    human.check_adult()
```
